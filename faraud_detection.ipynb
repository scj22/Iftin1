```python
new_df["isFlaggedFraud"]=new_df["isFlaggedFraud"].fillna(new_df["isFlaggedFraud"].mode()[0])
```

```python
new_df["newbalanceOrig"]=new_df["newbalanceOrig"].fillna(new_df["newbalanceOrig"].mean())
```

```python
new_df["nameDest"]=new_df["nameDest"].fillna(new_df["nameDest"].mode()[0])
```

```python
x=new_df.drop(columns=["isFraud"])
```

```python
x
```

```python
y=new_df['isFraud']
```

```python
y
```

```python
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder,StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.metrics import accuracy_score, precision_score, recall_score,f1_score,ConfusionMatrixDisplay
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
```

```python
x_train, x_text, y_train, y_test = train_test_split(
    x,
    y,
    random_state=42,
    test_size=0.2
)
```

```python
cat=[
    "type",
    "nameDest",
    "nameOrig"
]
num=[
    "step",
    "amount",
    "oldbalanceOrg",
    "newbalanceOrig",
    "oldbalanceDest",
    "newbalanceDest",
     "isFlaggedFraud"

]
```

```python
preprocesser=ColumnTransformer([
    ("cat",OneHotEncoder(handle_unknown='ignore'), cat),
    ("num",StandardScaler(),num)
 ]
)
```

```python
logistic= Pipeline(
    steps=[
        ("preprocessor",preprocesser),
        ("logistic",LogisticRegression())
    ]
)
```

```python
logistic.fit(x_train,y_train)
```

```python
y_pred=logistic.predict(x_text)
```

```python
accuracy=accuracy_score(y_test,y_pred)
precision_score=precision_score(y_test,y_pred)
recall_score=recall_score(y_test,y_pred)
f1_score=f1_score(y_test,y_pred)

print("Accuracy:",accuracy)
print("Precision Score:",precision_score)
print("Recall Score:",recall_score)
print("F1 Score:",f1_score)
```

```python
import joblib
joblib.dump(logistic,"logistic.joblib")
print("saved seccessfully")
```
